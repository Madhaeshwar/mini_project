```python
import os
import mysql.connector
from dotenv import load_dotenv
from pathlib import Path
```

```python
try:
    connection = mysql.connector.connect(
        host="localhost",
        user="root",
        password= os.getenv("URBAN_PULSE_DB_PASSWORD"),
        database="urban_pulse_db"
    )

    if connection.is_connected():
        print("MySQL database connection successful!")

except mysql.connector.Error as error:
    print("Database connection failed.")
    print("Error:", error)
```

```python
cursor = connection.cursor()

cursor.execute("SELECT COUNT(*) FROM service_requests")

total_records = cursor.fetchone()[0]

print("Total records in service_requests:", total_records)
```

```python
cursor.execute("DESCRIBE service_requests")

columns = cursor.fetchall()

for column in columns:
    print(column[0], "-", column[1])
```

```python
cursor.close()
connection.close()

print("Database connection closed successfully.")
```
